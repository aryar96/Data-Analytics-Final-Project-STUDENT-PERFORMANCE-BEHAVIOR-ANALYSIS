Replacing Missing Values with "Unknown" using "fillna"

```python
df["Parent_Education_Level"]=df["Parent_Education_Level"].fillna("Unknown")
```

For this dataset, since 1,025 out of 5,000 records (20.5%) have missing values in Parent_Education_Level, replacing all missing values with the mode
(Bachelor's) may introduce bias by assuming that these students belong to the most frequent category. Instead, treating the missing values as a 
separate Unknown category preserves the original information and clearly indicates that the parents' education level was not available. 
This approach allows us to retain all 5,000 records without making assumptions and provides a more transparent representation of the data for further
analysis.

4.4 Data Validation
</div>

```python
# Attendance validation-Attendance should be between 0 and 100%.
invalid_attendance = df[
    (df["Attendance (%)"] < 0) |
    (df["Attendance (%)"] > 100)
]
print("Invalid attendance records:", len(invalid_attendance))
```

```python
# Age validation-Age should be between 18 and 24.
invalid_age = df[
    (df["Age"] < 18) |
    (df["Age"] > 24)
]
print("Invalid age records:", len(invalid_age))
```

```python
#Student_ID Validation- Student_IDs should be unique.
print("Duplicate Student IDs:", df["Student_ID"].duplicated().sum())
```

```python
#Gender Validation- Gender data should be consistent
valid_gender = ["Male", "Female","Others"]

invalid_gender = df[
    ~df["Gender"].isin(valid_gender)
]

print("Invalid gender records:", len(invalid_gender))
```

```python
#Student score Validation-score columns should logically be within their intended scale(0-100)
score_columns = [
    "Midterm_Score",
    "Final_Score",
    "Assignments_Avg",
    "Quizzes_Avg",
    "Participation_Score",
    "Projects_Score",
    "Total_Score"
]

for col in score_columns:
    invalid_count = ((df[col] < 0) | (df[col] > 100)).sum()
    print(f"{col}: {invalid_count} invalid values")
```

```python
#Study hours validation-Study hours per week cannot logically be negative.
invalid_study_hours = df[
    df["Study_Hours_per_Week"] < 0
]

print("Invalid study-hour records:", len(invalid_study_hours))
```

```python
#Stress Level validation- Stress Level should be between 1 and 10
invalid_stress = df[
    (df["Stress_Level (1-10)"] < 1) |
    (df["Stress_Level (1-10)"] > 10)
]

print("Invalid stress-level records:", len(invalid_stress))
```

```python
#Sleep Hours validation-Sleep hours per night should not be negative
invalid_sleep_hours=df[
    (df["Sleep_Hours_per_Night"] <0)]

print("Invalid sleep hours records:",len(invalid_sleep_hours))
```

```python
#Department validation
df["Department"].value_counts(dropna=False)
valid_departments= [
    "Engineering",
    "Business",
    "CS",
    "Mathematics"]
invalid_departments=df[
    ~df["Department"].isin(valid_departments)]
print("Invalid department records:",len(invalid_departments))
```

```python
#Extracurricular Activities validation
invalid_Extracurricular_Activities = df[
    ~df["Extracurricular_Activities"].isin(["Yes", "No"])]
print("Invalid extracurricular Activities records:", len(invalid_Extracurricular_Activities))
```

```python
#Internet Access validation
invalid_internet_access = df[
    ~df["Internet_Access_at_Home"].isin(["Yes", "No"])]
print("Invalid internet access records:", len(invalid_internet_access))
```

```python
#Parent_Education_Level Validation
df["Parent_Education_Level"].value_counts()
valid_education = [
    "Bachelor's",
    "Master's",
    "PhD",
    "High School"
]
invalid_education = df[
    df["Parent_Education_Level"].notna() &
    ~df["Parent_Education_Level"].isin(valid_education)]
print("Invalid parent education records:", len(invalid_education))
```

```python
#Family_Income_Level validation
df["Family_Income_Level"].value_counts()
valid_income_level=[
    "Low",
    "Medium",
    "High"]
invalid_income_level=df[
    ~df["Family_Income_Level"].isin(valid_income_level)]
print("Invalid family income records:",len(invalid_income_level))
```

Data Validation

The dataset was validated to ensure that the values were within their expected ranges and that categorical variables contained consistent values. Numerical columns such as age, attendance, academic scores, study hours, stress level, and sleep hours were checked for invalid or unrealistic values. Categorical and derived columns were also reviewed for consistency. No invalid data was identified during the validation process, so no corrections or removals were required.

5.7c Quartile / IQR Analysis
</div>

Quartile and IQR analysis is a statistical method used to understand how the middle part of your data is distributed and to identify possible outliers

```python
# Calculate Q1 and Q3
Q1 = df["Total_Score"].quantile(0.25)
Q3 = df["Total_Score"].quantile(0.75)

# Calculate IQR (Interquartile Range)
IQR = Q3 - Q1

# Calculate lower and upper bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Identify outliers
outliers = df[
    (df["Total_Score"] < lower_bound) |
    (df["Total_Score"] > upper_bound)
]

# Display statistical values
print("Q1:", round(Q1, 2))
print("Q3:", round(Q3, 2))
print("IQR:", round(IQR, 2))
print("Lower Bound:", round(lower_bound, 2))
print("Upper Bound:", round(upper_bound, 2))
print("Number of Outliers:", len(outliers))
```

```python
# Display only student details and Total_Score comes under outlier
outliers[["Student_ID", "Full_Name", "Total_Score"]]
```

Quartile / IQR Analysis - Key Findings 

1) The middle 50% of Total_Score values range from 66.53 to 76.71
2) The IQR is 10.18, indicating the spread of the middle 50% of scores
3) The lower and upper outlier boundaries are 51.27 and 91.98
4) Only 4 observations were identified as potential outliers
5) Total_Score is relatively consistent, with most students' scores falling within the normal IQR range
6) These outliers were retained because they represent valid observations in the dataset
